```python
import pandas as pd
from graphrag_pipeline import GraphRAGQueryEngine
from langchain_community.chat_models.openai import ChatOpenAI
from langchain.schema import HumanMessage
from dotenv import load_dotenv
import os
```

```python
graphrag_engine = GraphRAGQueryEngine("../dataset/clean_fields_filter_date_sample50.csv")
# faiss_engine = FAISSSemanticQueryEngine("../dataset/clean_fields_filter_date_sample50.csv")
```

```python
load_dotenv()
openai_api_key = os.getenv("OPENAI_API_KEY")
if not openai_api_key:
    raise ValueError("OPENAI_API_KEY not set in your .env file.")

llm = ChatOpenAI(openai_api_key=openai_api_key, model="gpt-3.5-turbo", temperature=0, verbose=True)

def query_full_context(query_text: str, full_context: str) -> str:
    prompt = f"""
    You are an expert computer science research assistant.
    You are provided with one or more aggregated summaries of research paper abstracts.
    Even if these summaries come from different sets of papers, your task is to ignore their individual groupings
    and produce an answer to the following question. Account for the overall research themes, methodologies, significant findings,
    and emerging trends for the entire collection.
    \n\nAggregated Abstracts:\n\n
    {full_context}\n\n
    Question: {query_text}
    """
    
    # Generate response
    response = llm.invoke([HumanMessage(content=prompt)])
    return response.content
```

```python
# set up naive retrieval
df = pd.read_csv('../dataset/clean_fields_filter_date_sample50.csv')
df = df[df['abstract'].notna()]
df['abstract'] = df['abstract'].astype(str)
full_context = "\n\n".join(df['abstract'].fillna("").tolist())
```

```python
# naive retrieval test
naive_query_response = query_full_context(full_context=full_context, query_text = "what are the significant findings of AI papers before 2024?")
naive_query_response
```

```python
queries = [
    'What are the significant findings of AI papers before 2024?',
    'Summarize all machine learning papers',
    'Summarize machine learning papers after 2023',
    'What are the main themes of machine learning papers between 2021 and 2023?',
    'Find all papers from March 2023 to November 2023 on AI',
    'Summarize all papers from February 2025',
    'What are the main themes of papers from December 2024?',
    'What were the significant findings of papers in 2024?'
]
```

**Generate Naive and GraphRAG Responses to Queries:**

```python
responses_data = []

for i, query in enumerate(queries):
    naive_response = query_full_context(full_context=full_context, query_text=query)
    graphrag_response = graphrag_engine.query(query)
    responses_data.append({
        'query': query,
        'naive_response': naive_response,
        'graphrag_response': graphrag_response, 
        'semantic_cube_response': semantic_cube_responses[i]
    })

responses_df = pd.DataFrame(responses_data)
print(len(responses_df))
responses_df.head()
```

```python
responses_df.to_csv('responses.csv', index=False)
```

**LLM as a Judge Comparison for Naive, GraphRAG, SemanticCube:**

```python
from llm_judge_comparison import build_judge_prompt
```

```python
results = []
for i, row in responses_df.iterrows():
    judge_prompt = build_judge_prompt(row['query'], row['naive_response'], row['graphrag_response'], row['semantic_cube_response'])
    response = llm.invoke([HumanMessage(content=judge_prompt)])
    print(f"Query: {row['query']}")

    scores = eval(response.content)
    result_row = {
                "query": queries[0],
                **{f"naive_{k}": v for k, v in scores["naive"].items()},
                **{f"graphrag_{k}": v for k, v in scores["graphrag"].items()},
                **{f"cube_{k}": v for k, v in scores["cube"].items()}
            }

    results.append(result_row)

results_df = pd.DataFrame(results)
results_df.head()
```

```python
def calculate_average_scores(df: pd.DataFrame) -> pd.DataFrame:
    # List of criteria columns for each model
    criteria = ["comprehensiveness", "diversity", "directness", "empowerment"]
    
    # Prepare an empty list to store results
    aggregate_results = []

    # Iterate over each row (each query) in the original DataFrame
    for _, row in df.iterrows():
        # Calculate average score for each model (Naive, GraphRAG, Cube)
        naive_avg = sum(row[f"naive_{criterion}"] for criterion in criteria) / len(criteria)
        graphrag_avg = sum(row[f"graphrag_{criterion}"] for criterion in criteria) / len(criteria)
        cube_avg = sum(row[f"cube_{criterion}"] for criterion in criteria) / len(criteria)
        
        # Create a result dictionary with the query and average scores
        aggregate_results.append({
            "query": row["query"],
            "naive_avg_score": naive_avg,
            "graphrag_avg_score": graphrag_avg,
            "cube_avg_score": cube_avg
        })
    
    # Create a DataFrame from the results
    aggregate_df = pd.DataFrame(aggregate_results)
    return aggregate_df

# Example usage with your existing DataFrame (assuming df contains the scores)
aggregate_scores_df = calculate_average_scores(results_df)
aggregate_scores_df
```
